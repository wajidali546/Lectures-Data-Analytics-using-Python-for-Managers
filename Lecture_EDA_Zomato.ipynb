# Section 2: Summary Statistics (12 min)

Summary statistics describe a whole column with just a few numbers.

- **Centre**: mean, median, mode (where is the "middle"?)
- **Spread**: min, max, standard deviation, quartiles (how far apart are the values?)
- **Shape**: is the data balanced or stretched to one side? (compare mean and median)

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Summary statistics** | A few numbers that describe a whole column. | Mean, median, std |
| **Mean** | Sum of all values ÷ number of values. Affected by very large/small values. | Average order amount |
| **Median** | The middle value when sorted. Not affected much by extremes. | Typical order amount |
| **Mode** | The most frequent value. | Most common cuisine |
| **Minimum / Maximum** | Smallest / largest value. | Cheapest / biggest order |
| **Range** | Maximum − Minimum. | Spread of order amounts |
| **Standard deviation** | Average distance of values from the mean. Large = very spread out. | Order amounts vary by ~Rs 460 |
| **Quartiles (Q1, Q2, Q3)** | Values that split sorted data into 4 equal parts (25%, 50%, 75%). | Q3 = 75% of orders are below this |
| **IQR** | Interquartile Range = Q3 − Q1. The spread of the middle 50%. | Used to find outliers |
| **Skewness** | When data stretches to one side. Mean > median means right-skewed. | A few big orders |
| **`value_counts()`** | Counts how often each category appears. | Orders per city |

````python
# 2.1  describe(): one command, many statistics (numerical columns)
df[number_cols].describe().round(2)
````

````python
# 2.2  Statistics for one important column: Final_Amount (what the customer paid)
amount = df["Final_Amount"]

print("Mean (average)      : Rs", round(amount.mean(), 2))
print("Median (middle)     : Rs", amount.median())
print("Mode (most common)  : Rs", amount.mode()[0])
print("Minimum             : Rs", amount.min())
print("Maximum             : Rs", amount.max())
print("Standard deviation  : Rs", round(amount.std(), 2))
print("Range (max - min)   : Rs", amount.max() - amount.min())
````

````python
# 2.3  Mean vs Median: what does the gap tell us?
mean_value   = amount.mean()
median_value = amount.median()
print(f"Mean   = Rs {mean_value:,.2f}")
print(f"Median = Rs {median_value:,.2f}")

if mean_value > median_value:
    print("Mean is higher than median -> a few very large orders pull the average UP (right-skewed).")
else:
    print("Mean is close to or lower than median -> no strong pull from large orders.")
````

````python
# 2.4  Quartiles and IQR (used later for outliers)
q1 = amount.quantile(0.25)
q2 = amount.quantile(0.50)
q3 = amount.quantile(0.75)
iqr = q3 - q1

print("Q1 (25% of orders paid less than): Rs", q1)
print("Q2 (median)                       : Rs", q2)
print("Q3 (75% of orders paid less than): Rs", q3)
print("IQR (Q3 - Q1)                     : Rs", iqr)
````

````python
# 2.5  Categorical columns: count and percentage
print("Orders per City:")
print(df["City"].value_counts())

print("\nPayment mode share (%):")
print((df["Payment_Mode"].value_counts(normalize=True) * 100).round(1))

print("\nCuisine share (%):")
print((df["Cuisine"].value_counts(normalize=True) * 100).round(1))
````

````python
# 2.6  More quick facts
print("Number of cities   :", df["City"].nunique())
print("Number of cuisines :", df["Cuisine"].nunique())
print("Most common cuisine:", df["Cuisine"].mode()[0])
print("Average rating     :", round(df["Rating"].mean(), 2), "(blank ratings are ignored)")
print("Average delivery   :", round(df["Delivery_Time_Min"].mean(), 1), "minutes")
````

# Section 3: Group-wise Summaries using Pandas (15 min)

Overall numbers hide differences. **Group-wise summaries** split the data into groups (city, cuisine, month...) and calculate the same measure for each group.

```
groupby("City")  →  split into 5 city groups  →  calculate  →  combine into one table
```

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Group-wise summary** | Calculating a measure separately for each group. | Revenue per city |
| **`groupby()`** | Splits data into groups by a column. | `df.groupby("City")` |
| **Aggregation** | Reducing many values to one: sum, mean, count, min, max. | Total revenue of a city |
| **`agg()`** | Applies several aggregations at once, with named outputs. | Orders, revenue, avg rating |
| **`sort_values()`** | Arranges rows in order. | Cities by revenue |
| **`pivot_table()`** | Summary with one variable down the rows and another across the columns. | City × Cuisine average |
| **`crosstab()`** | A pivot table that counts occurrences. | Orders per City × Cuisine |
| **`idxmax()` / `idxmin()`** | The label (name) of the highest / lowest value. | City with highest revenue |
| **`reindex()`** | Puts rows into a chosen order. | July, August, September |
| **Revenue** | Money earned from sales. | Sum of `Final_Amount` |

````python
# 3.1  One group, one measure: total revenue per city
print(df.groupby("City")["Final_Amount"].sum())
````

````python
# 3.2  One group, many measures with agg()
city_summary = df.groupby("City").agg(
    Orders           = ("Order_ID", "count"),
    Revenue          = ("Final_Amount", "sum"),
    Avg_Order        = ("Final_Amount", "mean"),
    Avg_Delivery_Min = ("Delivery_Time_Min", "mean"),
    Avg_Rating       = ("Rating", "mean"),
).round(2)

city_summary = city_summary.sort_values("Revenue", ascending=False)
city_summary
````

````python
# 3.3  Cuisine summary
cuisine_summary = df.groupby("Cuisine").agg(
    Orders    = ("Order_ID", "count"),
    Revenue   = ("Final_Amount", "sum"),
    Avg_Order = ("Final_Amount", "mean"),
).round(2).sort_values("Revenue", ascending=False)

cuisine_summary
````

````python
# 3.4  Customer segment summary
segment_summary = df.groupby("Customer_Segment").agg(
    Orders       = ("Order_ID", "count"),
    Avg_Order    = ("Final_Amount", "mean"),
    Avg_Discount = ("Discount_Pct", "mean"),
    Avg_Rating   = ("Rating", "mean"),
).round(2)

segment_summary
````

````python
# 3.5  Time summary: month and weekday
monthly = df.groupby("Month").agg(
    Orders  = ("Order_ID", "count"),
    Revenue = ("Final_Amount", "sum"),
).reindex(month_order)                       # keep months in calendar order

print(monthly)

print("\nOrders by weekday:")
print(df.groupby("Weekday")["Order_ID"].count().reindex(day_order))
````

````python
# 3.6  Two groups together: pivot_table (City x Cuisine = average order value)
pivot_avg = df.pivot_table(values="Final_Amount", index="City", columns="Cuisine", aggfunc="mean").round(0)
pivot_avg
````

````python
# 3.7  Count table: crosstab (City x Cuisine = number of orders). Look for zeros!
city_cuisine = pd.crosstab(df["City"], df["Cuisine"])
city_cuisine
````

````python
# 3.8  Ask the data direct questions
print("City with highest revenue :", city_summary["Revenue"].idxmax())
print("City with slowest delivery:", city_summary["Avg_Delivery_Min"].idxmax())
print("City with best rating     :", city_summary["Avg_Rating"].idxmax())
print("Cuisine with most revenue :", cuisine_summary["Revenue"].idxmax())
print("Cuisine with highest avg order:", cuisine_summary["Avg_Order"].idxmax())
````

# Section 4: Simple Plots for EDA (20 min)

A plot shows in seconds what a table needs minutes to explain. We use **four libraries**, so students see that the same ideas work in every tool.

### Which chart for which question?
| Question | Chart | Variable types |
|---|---|---|
| How are values **distributed**? | Histogram | 1 number |
| Is there an **outlier / spread**? | Boxplot | 1 number (by group) |
| How do **groups compare**? | Bar chart | category + number |
| How does it **change over time**? | Line chart | date/month + number |
| Are **two numbers related**? | Scatter plot | number + number |
| Where are the **gaps / hot spots**? | Heatmap | category × category |

### Libraries
| Library | Strength | Style |
|---|---|---|
| Pandas `.plot()` | Fastest, one line | Quick look |
| Matplotlib | Full control, the base of most others | Basic |
| Seaborn | Beautiful statistical charts | Easy and clean |
| Plotly | Interactive: hover, zoom, click | Dashboards |

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Visualisation** | Showing data as a picture so patterns are easy to see. | Any chart |
| **Histogram** | Bars showing how many values fall in each range. Shows **distribution**. | Spread of order amounts |
| **Boxplot** | Box = middle 50%, line = median, dots = outliers. Shows **spread and outliers**. | Delivery time by city |
| **Bar chart** | One bar per category. Compares **groups**. | Revenue by city |
| **Line chart** | Points joined in time order. Shows **trend**. | Orders per month |
| **Scatter plot** | One dot per record on two number axes. Shows **relationship**. | Distance vs delivery time |
| **Heatmap** | A table coloured by value. Shows **hot spots and gaps**. | City × Cuisine counts |
| **Matplotlib** | The base plotting library with full control. | `plt.bar()` |
| **Seaborn** | A library built on Matplotlib for clean statistical charts. | `sns.boxplot()` |
| **Plotly** | A library for interactive charts (hover, zoom). | `px.scatter()` |
| **Axis / Title / Label** | The scale lines, heading and names of a chart. Always add them. | `plt.title(...)` |

# Section 5: Patterns, Gaps, Trends and Outliers (10 min)

Four things to look for in every dataset:

| Look for | Question | Tools |
|---|---|---|
| **Pattern** | Do two things move together? Do groups differ? | `corr`, `groupby`, scatter |
| **Gap** | What is **missing** or unusually low? | `crosstab`, heatmap, zeros |
| **Trend** | How does it change over time? | monthly `groupby`, line chart |
| **Outlier** | What is **very different** from the rest? | IQR rule, boxplot |

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Pattern** | A regular relationship or difference in the data. | Longer distance, longer delivery |
| **Correlation** | A number from −1 to +1 showing how two variables move together. | Distance vs time = 0.79 |
| **Positive / Negative correlation** | Both rise together / one rises while the other falls. | Time vs rating is negative |
| **Gap** | A missing or unusually low area. | 0 South Indian orders in Delhi |
| **Trend** | A direction of change over time. | Orders rising Jul → Sep |
| **Growth %** | (New − Old) ÷ Old × 100. `pct_change()` calculates it. | July to August growth |
| **Outlier** | A value far away from the rest. | A Rs 4,365 order |
| **IQR rule** | Outlier if above Q3 + 1.5×IQR or below Q1 − 1.5×IQR. | Fence for Final_Amount |
| **Binning (`pd.cut`)** | Grouping numbers into ranges. | Distance bands |
| **Causation** | One thing *makes* another happen. Correlation alone cannot prove it. | Distance may cause delay, but rain? Unknown |

````python
# 5.1  PATTERN: does a longer distance mean a longer delivery time?
corr = df["Distance_Km"].corr(df["Delivery_Time_Min"])
print("Correlation (distance vs delivery time):", round(corr, 2))

# Group distance into bands
df["Distance_Band"] = pd.cut(df["Distance_Km"], bins=[0, 3, 6, 9, 12],
                             labels=["0-3 km", "3-6 km", "6-9 km", "9-12 km"])
print("\nAverage delivery time by distance band:")
print(df.groupby("Distance_Band", observed=True)["Delivery_Time_Min"].mean().round(1))
````

````python
# 5.2  PATTERN: does a slow delivery lead to a lower rating?
print("Correlation (delivery time vs rating):", round(df["Delivery_Time_Min"].corr(df["Rating"]), 2))

df["Time_Band"] = pd.cut(df["Delivery_Time_Min"], bins=[0, 30, 45, 60, 200],
                         labels=["Up to 30 min", "31-45 min", "46-60 min", "Over 60 min"])
print("\nAverage rating by delivery time band:")
print(df.groupby("Time_Band", observed=True)["Rating"].mean().round(2))
````

````python
# 5.3  GAP: which City-Cuisine combinations have ZERO orders?
cells = city_cuisine.stack()               # turn the table into one long list
print("Combinations with zero orders:")
print(cells[cells == 0])

print("\nThe 5 thinnest combinations (fewest orders):")
print(cells.sort_values().head(5))
````

````python
# 5.4  TREND: month-by-month growth
monthly["Orders_Growth_%"]  = (monthly["Orders"].pct_change() * 100).round(1)
monthly["Revenue_Growth_%"] = (monthly["Revenue"].pct_change() * 100).round(1)
monthly
````

````python
# 5.5  OUTLIERS: the IQR rule.  Outlier = above Q3 + 1.5 x IQR   (or below Q1 - 1.5 x IQR)
for col in ["Final_Amount", "Delivery_Time_Min"]:
    q1  = df[col].quantile(0.25)
    q3  = df[col].quantile(0.75)
    iqr = q3 - q1
    upper = q3 + 1.5 * iqr
    lower = q1 - 1.5 * iqr
    n_out = ((df[col] > upper) | (df[col] < lower)).sum()
    print(f"{col}: normal range = {lower:.1f} to {upper:.1f}  ->  {n_out} outliers")
````

````python
# 5.6  Look at the most extreme orders
cols = ["Order_ID", "City", "Cuisine", "Items_Count", "Final_Amount", "Delivery_Time_Min", "Rating"]

print("Top 3 largest orders:")
print(df.nlargest(3, "Final_Amount")[cols])

print("\nTop 3 slowest deliveries:")
print(df.nlargest(3, "Delivery_Time_Min")[cols])
````

````python
# 5.7  How much does one outlier change the average?
with_outlier    = df["Final_Amount"].mean()
without_outlier = df[df["Final_Amount"] < df["Final_Amount"].max()]["Final_Amount"].mean()

print(f"Average order WITH the biggest order   : Rs {with_outlier:,.2f}")
print(f"Average order WITHOUT the biggest order: Rs {without_outlier:,.2f}")
print(f"Median order (not affected much)       : Rs {df['Final_Amount'].median():,.2f}")
````

# Section 6: Connecting Findings to Business Meaning (5 min)

A number alone is not an insight. Every finding must answer three questions:

```
WHAT does the data show?   →   WHY might it matter?   →   WHAT could the business do?
```

| Finding type | Ask yourself | Example meaning |
|---|---|---|
| **Pattern** | Is this a cause or just a link? | Longer distance, longer delivery: delivery radius may matter |
| **Gap** | Is it a real opportunity or a data problem? | No orders: restaurant not available, or demand not reached? |
| **Trend** | Will it continue? Can operations cope? | Fast growth: more riders and servers needed |
| **Outlier** | Error, or a special case? | One huge order: party/bulk order, not a normal customer |

> ⚠️ **Correlation is not causation.** Our data shows *what* happens, not always *why*. Reasons are **possible explanations to be tested**.

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Business meaning** | What a finding implies for decisions and money. | Slow delivery may lower ratings |
| **Insight** | A finding that explains *why it matters* and points to an *action*. | "Add riders in far zones" |
| **Hypothesis** | A possible explanation that still needs testing. | "Traffic may slow Bengaluru" |
| **Correlation vs causation** | Moving together is not the same as one causing the other. | Discount and revenue |
| **Actionable** | Something a manager can actually do. | Onboard Delhi restaurants |
| **KPI** | Key Performance Indicator: the number a business tracks. | Avg delivery time |
| **Sample limit** | Results come from 200 orders, not all orders. | Careful about generalising |

````python
# Build a findings table using REAL numbers from our outputs
slow_city   = city_summary["Avg_Delivery_Min"].idxmax()
slow_time   = city_summary["Avg_Delivery_Min"].max()
fast_time   = city_summary["Avg_Delivery_Min"].min()
sep_orders  = monthly.loc["September", "Orders"]
jul_orders  = monthly.loc["July", "Orders"]
biggest     = df["Final_Amount"].max()
corr_dist   = df["Distance_Km"].corr(df["Delivery_Time_Min"])
rating_slow = df[df["Delivery_Time_Min"] > 60]["Rating"].mean()
rating_fast = df[df["Delivery_Time_Min"] <= 30]["Rating"].mean()

findings = pd.DataFrame({
    "Type": ["Pattern", "Pattern", "Gap", "Trend", "Outlier"],
    "What the data shows": [
        f"Distance and delivery time move together (correlation {corr_dist:.2f})",
        f"Orders over 60 min average rating {rating_slow:.1f} vs {rating_fast:.1f} for orders up to 30 min",
        "Delhi has 0 South Indian orders; other cities have a few",
        f"Orders grew from {jul_orders} (July) to {sep_orders} (September)",
        f"One order of Rs {biggest:,.0f} is far above normal",
    ],
    "Possible business meaning": [
        "Long-distance orders need more riding time; delivery radius affects service quality",
        "Slow delivery may hurt customer satisfaction",
        "No restaurant partners, low visibility, or low demand in Delhi",
        "Platform demand is rising; rider and server capacity must keep up",
        "Likely a party/bulk order, so it should not be treated as a normal customer",
    ],
    "Possible action": [
        "Limit radius or add riders in far zones",
        "Track and reduce orders above 60 min",
        "Check partner list; test a South Indian campaign in Delhi",
        "Plan rider hiring before peak weeks",
        "Report median alongside mean; treat bulk orders separately",
    ],
})
findings
````

# Section 7: AI-Assisted EDA Observation Statements (6 min)

**Observation statement** = a short, factual sentence describing what the data shows.

### The 3-part formula
```
Observation  =  METRIC  +  COMPARISON  +  EVIDENCE (the number)
```

| Weak | Strong |
|---|---|
| "Bengaluru is slow." | "Bengaluru has the slowest average delivery (42.7 min), 7 min longer than Delhi (35.7 min)." |
| "Orders are growing." | "Orders rose from 51 in July to 86 in September (+69%)." |
| "Some outliers exist." | "One order of Rs 4,365 is almost 9 times the median order (Rs 500)." |

### How AI helps, and its risks
| AI is good at | AI can go wrong by |
|---|---|
| Drafting sentences quickly | Inventing or changing numbers |
| Finding comparisons in a table | Mixing up which group is highest |
| Using clear business language | Claiming causes the data cannot prove |

### Good prompt = Role + Context + Data + Task + Rules

### 📖 Definitions
| Term | Definition | Zomato example |

| **AI-assisted EDA** | Using an AI assistant to help describe results, while the analyst checks them. | Draft observations |
| **Prompt** | The instruction text you give to an AI. | The prompt in this section |
| **Prompt formula** | Role + Context + Data + Task + Rules. | Used above |

# Section 8: AI-Supported Managerial Insight Writing (5 min)

An **observation** says *what*. An **insight** tells a manager *what it means and what to do*.

### The insight structure
| Part | Question it answers | Example |
|---|---|---|
| **Finding** | What does the data show? | Bengaluru deliveries average 42.7 min |
| **Why it matters** | So what? | Slower orders get lower ratings |
| **Action** | What should we do? | Add riders in Bengaluru's far zones |
| **Metric to track** | How will we know it worked? | Average delivery time, share of orders over 60 min |
| **Caution** | What do we not know? | Data has no traffic or weather information |

### Writing tips for managers
| Do | Don't |
|---|---|
| Lead with the most important point | Start with technical steps |
| Use 1-2 numbers per sentence | Paste a table of 50 numbers |
| Say "may" when giving reasons | State causes as facts |
| End with a clear action | End with only a description |

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Managerial insight** | A finding written for a manager, with meaning and action. | The manager brief |
| **Finding** | What the data shows. | Bengaluru averages 42.7 min |
| **Recommendation / Action** | What the business should do next. | Add riders |
| **Metric to track** | The measure that shows if the action worked. | Share of orders above 60 min |
| **Caution / Limitation** | What the data cannot tell us. | No weather data |
| **Stakeholder** | The person who uses the analysis. | City operations manager |
| **Hedging words** | Words like *may* or *could* used when a cause is not proven. | "may be due to traffic" |

````python
# Prompt to turn verified observations into a manager insight
obs_text = "\n".join(f"- {o}" for o in obs)

insight_prompt = f'''
ROLE: You are a business analyst writing for a city operations manager at a food delivery company.

VERIFIED OBSERVATIONS:
{obs_text}

TASK: Write a short managerial brief with 3 insights.
For each insight give: Finding, Why it matters, Recommended action, Metric to track.

RULES:
1. Use only the numbers above.
2. Use "may" or "could" for reasons, because the data cannot prove causes.
3. Maximum 120 words per insight. Simple business language.
'''
print(insight_prompt)
````

# Section 9: Verification using Actual Python Outputs (7 min)

> **Golden rule: AI helps you write. Python decides what is true.**

### Common AI mistakes
| Mistake | Example |
|---|---|
| **Wrong number** | Says Rs 520 when the average is Rs 607 |
| **Wrong ranking** | Says Delhi is highest when Bengaluru is |
| **Wrong group** | Mixes up the highest and the lowest |
| **Invented fact** | Says "no missing values" without checking |
| **Unsupported cause** | "Rain caused delays" when there is no weather data |

### Verification steps
1. Split the AI text into **single claims**.
2. Recompute each claim with Pandas.
3. Mark it: ✔ Correct, ✘ Incorrect, or ? Cannot verify with this data.
4. Rewrite the incorrect claims with the true numbers.

Below is an **AI-written draft** (it contains deliberate mistakes). Verify it.

### 📖 Definitions
| Term | Definition | Zomato example |
|---|---|---|
| **Verification** | Checking a statement against real data before trusting it. | Recompute the average |
| **Claim** | One single checkable statement. | "UPI is 48% of orders" |
| **Tolerance** | How small a difference is still accepted (rounding). | 607 vs 606.6 |
| **Correct / Incorrect / Cannot verify** | The three possible results of a check. | ✔ ✘ ? |
| **Audit trail** | Keeping the code that produced each number. | The `check()` cells |
| **Human-in-the-loop** | A person reviews and approves AI output before use. | The analyst |
| **Over-claiming** | Stating more than the data can support. | "Rain caused delays" |

````python
ai_draft = '''
AI DRAFT SUMMARY (to be verified)
1. The average order amount is about Rs 607.
2. Mumbai has the lowest average order amount.
3. Orders grew every month, reaching 86 in September.
4. Biryani earns the most revenue of all cuisines.
5. Delhi has the slowest average delivery time.
6. UPI is used in about 48% of orders.
7. The dataset has no missing values.
8. Slow deliveries in Bengaluru are caused by rain.
9. Delhi has no South Indian orders.
'''
print(ai_draft)
````

````python
# A small helper: compare what AI claimed with what Python found
def check(number, claim, ai_value, python_value, tolerance=0):
    if isinstance(python_value, (int, float, np.number)):
        ok = abs(ai_value - python_value) <= tolerance
    else:
        ok = (ai_value == python_value)
    status = "✔ CORRECT  " if ok else "✘ INCORRECT"
    print(f"{status} | Claim {number}: {claim}")
    print(f"             AI said: {ai_value}   |   Python found: {python_value}")

# --- Python values for each claim ---
avg_amount  = round(df["Final_Amount"].mean(), 1)
lowest_avg  = city_summary["Avg_Order"].idxmin()
sep_count   = int(monthly.loc["September", "Orders"])
grew_always = bool(monthly["Orders"].is_monotonic_increasing)
top_cuisine = cuisine_summary["Revenue"].idxmax()
slowest     = city_summary["Avg_Delivery_Min"].idxmax()
upi_pct     = round((df["Payment_Mode"] == "UPI").mean() * 100, 1)
n_missing   = int(df.isnull().sum().sum())
delhi_south = int(city_cuisine.loc["Delhi", "South Indian"])

check(1, "Average order amount",            607,      avg_amount, tolerance=1)
check(2, "City with the lowest avg order",  "Mumbai", lowest_avg)
check(3, "September orders",                86,       sep_count)
check(3, "Orders grew every month",         True,     grew_always)
check(4, "Cuisine with most revenue",       "Biryani", top_cuisine)
check(5, "City with slowest delivery",      "Delhi",  slowest)
check(6, "UPI share (%)",                   48,       upi_pct, tolerance=0.5)
check(7, "Number of missing values",        0,        n_missing)
check(9, "Delhi South Indian orders",       0,        delhi_south)
````

````python
# Claim 8 cannot be checked with Python: our data has NO weather column.
print("? CANNOT VERIFY | Claim 8: 'Slow deliveries in Bengaluru are caused by rain'")
print("   Our columns:", list(df.columns[:15]))
print("   There is no weather or traffic variable, so a cause cannot be proven.")
print("   Safe wording: Bengaluru deliveries are slower (", round(city_summary.loc['Bengaluru', 'Avg_Delivery_Min'], 1),
      "min). The cause needs more data (for example traffic or weather).")
````

````python
# Final corrected version: every number now comes straight from Python
fixed = f'''
VERIFIED SUMMARY
1. The average order amount is about Rs {avg_amount:,.0f}.                                  (AI was right)
2. {lowest_avg} has the lowest average order amount (Rs {city_summary['Avg_Order'].min():,.0f}); {city_summary['Avg_Order'].idxmax()} has the highest (Rs {city_summary['Avg_Order'].max():,.0f}).
3. Orders grew every month, reaching {sep_count} in September.                               (AI was right)
4. {top_cuisine} earns the most revenue (Rs {cuisine_summary['Revenue'].max():,.0f}); Biryani ranks number {list(cuisine_summary.index).index('Biryani') + 1} by revenue.
5. {slowest} has the slowest average delivery time ({city_summary['Avg_Delivery_Min'].max():.1f} min).
6. UPI is used in {upi_pct}% of orders.                                                      (AI was right)
7. The dataset has {n_missing} missing values, all in the Rating column.
8. Bengaluru deliveries are slow, but the cause cannot be proven with this data.
9. Delhi has {delhi_south} South Indian orders.                                              (AI was right)
'''
print(fixed)
````
